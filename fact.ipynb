```python
# this notebook is for the FACT course by Fernando Santos at University of Amsterdam
# the notebook illustrates a Differentially Private algorithm to find the population mean under Local Differential Privacy

# The first cell is a singular estimate, as it would occur in practise
# The second cell is a Monte Carlo estimates to show the accuracy of the estimate
```

```python
import numpy as np
import matplotlib.pyplot as plt

num_samples = 300
true_rate = 0.71

data = np.random.choice([0, 1], size=num_samples, p=[1-true_rate, true_rate])
dice = np.random.randint(0, 3, size=num_samples)

outcome = np.where(dice <= 1, dice, data)
print(f"Estimate of outcome: {np.mean(outcome):.3f}")

rate_estimate = (np.mean(outcome) - 1/3) * 3
print(f"Estimate of rate:    {rate_estimate:.3f}")

print(f"True rate:           {true_rate:.3f}")

```

```python
num_mc = 50000

estimates = []
num_samples = 300
true_rate = 0.71
for i in range(num_mc):

  data = np.random.choice([0, 1], size=num_samples, p=[1-true_rate, true_rate])

  dice = np.random.randint(0, 3, size=num_samples)
  outcome = np.where(dice <= 1, dice, data)


  estimates.append((np.mean(outcome) - 1/3) * 3)
```

```python
plt.hist(estimates, bins=30);
```
